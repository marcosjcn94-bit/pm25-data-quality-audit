# Monthly PM2.5 sensitivity audit
Recompute the frozen tables from EPA daily summaries. Each row represents one monitor on one local calendar date. The study uses four monitors in Los Angeles County in 2024.

```python
from pathlib import Path
from tempfile import TemporaryDirectory
import pandas as pd
from IPython.display import display, Image
from io import BytesIO
from audit import run
from figure import plot

with TemporaryDirectory() as tmp:
    directory = Path(tmp)
    run(Path('data/daily.csv'), directory)
    monitor = pd.read_csv(directory / 'monitor_months.csv')
    monthly = pd.read_csv(directory / 'monthly.csv')
    image = BytesIO()
    plot(directory / 'monthly.csv').savefig(image, format='png', dpi=120)
```

## Eligibility and weighting
Count eligible monitor-months under each coverage rule, then inspect December, where the coverage filter changed the set of included monitors.

```python
display(monitor[['eligible_any','eligible_75','eligible_90']].sum())
display(monthly[monthly.month == '2024-12'])
display(Image(data=image.getvalue()))
```

The four-monitor mean describes the eligible monitors in the extract. Its change with the inclusion rule measures sensitivity to the set of monitors represented.
